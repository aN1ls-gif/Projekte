```python
import numpy as np
import matplotlib.pyplot as plt

def NormalDist(Mean, Std, shift):
    x = np.linspace(-2, 2, 71)
    y = (Std * np.sqrt(2*np.pi))**-1
    y *= np.exp( -1 * (np.square(x - Mean)/np.square(2*Std) ))
    return x, y

x, y = NormalDist(0, 0.3955)
x += 0.7
plt.plot(x, y)
plt.show()
```

```python
def SingularPointFromNormalDist(x, Mean, Std):
    y = (Std * np.sqrt(2*np.pi))**-1
    y *= np.exp( -1 * (np.square(x - Mean)/np.square(2*Std) ))
    return y

x = np.linspace(-2, 2, 71)
y = [SingularPointFromNormalDist(i, Mean = 0.7, Std = 0.399) for i in x]
plt.plot(x, y)
plt.show()
```
